# Часть 5

🐾 Теперь о KPI. Все вводные данные взяты с этих источников:

        CPC: https://texterra.ru/blog/skolko-stoit-reklama-v-vk.html,

        Conversion: https://pbd.space/blog/marketingovyye-pokazateli-effektivnosti/,

        Repeat rate: https://guldog.ru/perederjka,

```python
budget = 250000

cpc = 20
conversion = 0.24
repeat_rate = 0.40

#метрики
visitors = budget/cpc
new_users = int(visitors * conversion)
repeat_users = int(new_users * repeat_rate)
orders = (new_users + repeat_users)
CAC = budget/new_users

print("Ожидаемые KPI в месяц:\n")

print("Переходы на сайт:", round(visitors))
print("Новые пользователи:", new_users)
print("Повторные клиенты:", repeat_users)
print("Заказы:", orders)
print("CAC:", round(CAC))
```

🐾 С учетом того, что наша компания новая и определены не все цены на наши услуги, а разработана только примерная модель, предполагается, что примерные KPI, которые возможно расчитать будут следующими:

> **Переходы на сайт: 12500**

> **Новые пользователи: 3000**

> **Повторные клиенты: 1200**

> **Заказы: 4200**

> **CAC: 83 руб**
